```python
#Experiment 1: Weather Dataset Download and Loading Using Python
import pandas as pd

#Load weather dataset from Github
url = "https://raw.githubusercontent.com/arnabDAanalyst/Weather-and-Air-Quality-of-Indian-Cities-Analysis/refs/heads/main/indian_weather_data.csv"

df = pd.read_csv(url)

#Display first five rows
print("WEATHER DATASET")
print(df.head())

#Display dataset information
print("\nDataset Shape:",df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
#Experiment 4: Spam/Ham Filtering Using Machine Learning
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

messages = [
    "You have won a free iPhone",
    "Click here to claim your prize",
    "Earn money from home",
    "Congratulations you won a gift",
    "Good morning have a nice day",
    "Can we meet after college",
    "Please send me your notes",
    "I will call you in the evening"
]

labels = [
    "Spam", "Spam", "Spam", "Spam",
    "Ham", "Ham", "Ham", "Ham"
]

cv = CountVectorizer()
X = cv.fit_transform(messages)

model = MultinomialNB()
model.fit(X, labels)

test_messages = [
    "Claim your free gift",
    "Please send your notes",
    "Earn money now",
    "Good morning friend"
]

X_test = cv.transform(test_messages)
predictions = model.predict(X_test)

print("Spam and Ham Classification Results")
print("-----------------------------------")

for msg, result in zip(test_messages, predictions):
    print(msg, " --> ", result)
```
